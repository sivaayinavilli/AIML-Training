## Python Fundamentals for AI/ML

```python
# Task 1.1: Variables, data types, operators
a, b = 10, 3
print(a + b, a - b, a * b, a / b, a // b, a % b, a ** b)
```

```python
# Task 1.2: Conditional statements
def grade(m):
    if m >= 90:
        return "A"
    elif m >= 75:
        return "B"
    elif m >= 60:
        return "C"
    else:
        return "F"

print([grade(x) for x in (95, 80, 65, 40)])

n = 7
print("even" if n % 2 == 0 else "odd")
```

```python
# Task 1.3: Loops
for i in range(1, 4):
    print(i)
print("Sum 1 to 5:", sum(range(1, 6)))
```

```python
# while loop
count = 3
while count > 0:
    print(count)
    count -= 1
```

```python
#break and continue
for i in range(1, 10):
    if i == 3:
        continue      # skip 3
    if i == 6:
        break         # stop at 6
    print(i, end=" ")
```

```python
# Task 1.4: Collections
# List
L = [10, 20, 30]
L.append(40)
L.insert(1, 15)
L.remove(20)
print(L, L[-1], L[1:3], len(L))
```

```python
# Tuple
t = (3, 4)
x, y = t          # unpacking
print(x, y)
```

```python
# set
s = {1, 2, 2, 3}
s.add(4)
print(s)
print({1, 2, 3} & {2, 3, 4})   # intersection
print({1, 2} | {2, 3})          # union
print({1, 2, 3} - {2})          # difference
```

```python
# Dictionary
d = {"name": "Siva", "age": 22}
d["city"] = "Tuni"
for k, v in d.items():
    print(k, v)
print(d.get("salary", "N/A"), list(d.keys()))
```

```python
# Tasks 1.5 and 1.6: Functions and parameters
# Basic function
def mean(values):
    return sum(values) / len(values)

print(mean([70, 80, 90]))
```

```python
# Default, keyword, *args, **kwargs
def power(base, exp=2):            # exp has a default value
    return base ** exp

print(power(3))                    # positional -> 9
print(power(3, 3))                 # positional -> 27
print(power(exp=4, base=2))        # keyword    -> 16

def total(*nums):                  # any number of arguments
    return sum(nums)

def info(**kw):                    # any number of key=value pairs
    return kw

print(total(1, 2, 3), info(a=1, b=2))
```

```python
# Return multiple values
def stats(v):
    return min(v), max(v), mean(v)

lo, hi, av = stats([4, 8, 15])
print(lo, hi, av)
```

```python
# Task 1.7: List comprehension
print([x**2 for x in range(1, 6)])
print([x for x in range(10) if x % 2 == 0])
print(["even" if x % 2 == 0 else "odd" for x in range(4)])
print({w: len(w) for w in ["ai", "ml", "dl"]})   # dict comprehension
```

```python
# Task 1.8: Exception handling
# try, except, else, finally
def safe_div(a, b):
    try:
        r = a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"
    except TypeError:
        return "Numbers only"
    else:
        return r
    finally:
        print("done")           # always runs

print(safe_div(10, 2))
print(safe_div(5, 0))
print(safe_div("a", 2))
```

```python
# Task 1.9: File handling (text and CSV)
with open("notes.txt", "w") as f:       # "w" overwrites
    f.write("Line one\nLine two\n")

with open("notes.txt", "a") as f:       # "a" appends
    f.write("Line three\n")

with open("notes.txt") as f:            # default mode is read
    print(f.read())
```

```python
# Write a CSV file
import csv

rows = [["name", "marks"], ["Asha", 85], ["Ravi", 72], ["Meena", 91]]
with open("students.csv", "w", newline="") as f:
    csv.writer(f).writerows(rows)
```

```python
# Read a CSV file
with open("students.csv") as f:
    for r in csv.DictReader(f):
        print(r["name"], int(r["marks"]))
```

```python
# Read and write CSV with Pandas
import pandas as pd

df = pd.read_csv("students.csv")
df["result"] = df["marks"].apply(lambda m: "Pass" if m >= 75 else "Fail")
df.to_csv("students_result.csv", index=False)
df
```
