## Task 1.10: Ten coding exercises

```python
# 1. Print even numbers from 1 to 20.
print([n for n in range(1, 21) if n % 2 == 0])
```

```python
# 2. Factorial of a number.
def fact(n):
    r = 1
    for i in range(2, n + 1):
        r *= i
    return r

print(fact(5))          # 120
```

```python
# 3. Prime numbers up to 30.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print([n for n in range(1, 31) if is_prime(n)])
```

```python
# 4. Reverse a string and check for a palindrome.
print("machine"[::-1])              # enihcam
print("madam" == "madam"[::-1])     # True
```

```python
# 5. Count vowels in a sentence.
print(sum(1 for ch in "machine learning" if ch in "aeiou"))   # 6
```

```python
# 6. Word frequency.
from collections import Counter
print(dict(Counter("ai ml ai dl ml ai".split())))
```

```python
# 7. Largest and second largest number.
nums = [5, 12, 7, 20, 3]
print(max(nums), sorted(nums, reverse=True)[1])    # 20 12
```

```python
# 8. Remove duplicates but keep the order.
print(list(dict.fromkeys([1, 2, 2, 3, 1])))        # [1, 2, 3]
```

```python
# 9. FizzBuzz for 1 to 15.
for i in range(1, 16):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 10. Celsius to Fahrenheit converter.
def c2f(c):
    return c * 9 / 5 + 32

print(c2f(100), c2f(37))     # 212.0 98.6
```
