# Complete ML Workflow: Iris Flower Classification


## 1. Problem Definition
Predict the species of an iris flower (setosa, versicolor or virginica) from four flower measurements. This is a **classification** problem, because the output is a category.

## 2. Dataset
- **Source:** Iris dataset (built into scikit-learn)
- **Size:** 150 rows and 4 feature columns, plus a target
- **Features (X):** sepal length, sepal width, petal length, petal width (all in cm)
- **Target (y):** species (0 = setosa, 1 = versicolor, 2 = virginica)
- **Class balance:** 50 rows per species

## 3. Data Loading
Loaded with `load_iris(as_frame=True)` into a Pandas DataFrame, and species names were added for readability.

## 4. Data Cleaning
- Missing values: **0**
- Duplicate rows: **1** (rows 101 and 142, identical virginica). Harmless, so it was kept.
- Data types: all features are float, so no conversion was needed.
- Outliers: **4** in sepal width. They are natural variation, so they were kept.

## 5. Exploratory Data Analysis
- Petal length and petal width are the most useful features. Their averages rise clearly from setosa (1.46 / 0.25) to versicolor (4.26 / 1.33) to virginica (5.55 / 2.03).
- Petal length and petal width are strongly correlated (0.96).
- Setosa is fully separable from the others by petal length (1.0-1.9 cm vs 3.0+ cm).
- Versicolor and virginica overlap slightly, so they are the hardest pair to tell apart.

## 6. Visualizations
Histogram (petal length), box plot (by species), scatter plot (petal length vs width), correlation heatmap, and pair plot.

## 7. Train/Test Split
- 80% train (120 rows) and 20% test (30 rows)
- `random_state=42` for a repeatable split
- `stratify=y` to keep classes equal (40 per class in training, 10 per class in test)

## 8. Preprocessing
Applied `StandardScaler`. It was **fitted on the training data only** and then applied to the test data, to avoid data leakage.

## 9. Model Training
Trained a k-Nearest Neighbors classifier (k = 5). Also compared Logistic Regression and Decision Tree.

## 10. Evaluation
| Model | Test accuracy |
|---|---|
| kNN (k=5) | 0.933 |
| Logistic Regression | 0.933 |
| Decision Tree | 0.933 |

- kNN train accuracy was 0.975, close to test accuracy, so there is little overfitting.
- The confusion matrix shows every setosa and versicolor flower correct, and 2 virginica flowers misclassified as versicolor.

## 11. Prediction
A new flower with measurements (5.1, 3.5, 1.4, 0.2) was predicted as **setosa**.

## 12. Conclusion
The kNN model classified about 93% of test flowers correctly. The only errors were between versicolor and virginica, which matches the overlap seen in the EDA. Next steps: cross-validation, hyperparameter tuning and trying more algorithms.

## 13. Limitations
The test set is only 30 rows, so one flower changes the score by 3.3 points. The dataset is small and clean, so real-world data would be harder.
