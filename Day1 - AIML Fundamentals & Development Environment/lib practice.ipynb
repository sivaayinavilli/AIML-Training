Arithmetic

```python
a = 10
b = 3
print(a + b, a - b, a * b, a / b, a // b, a % b, a ** b)
```

Filter a list

```python
temps = [32, 35, 29, 40, 31]
hot = [t for t in temps if t > 32]
print(hot)
```

Count words with a dictionary

```python
text = "ai ml ai dl ml ai"
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

Your own function

```python
def average(values):
    return sum(values) / len(values)

print(average([70, 80, 90]))
```

## NumPy

Basic statistics

```python
import numpy as np

arr = np.array([4, 8, 15, 16, 23, 42])
print(arr.mean())
print(round(arr.std(), 2))
print(arr.min(), arr.max())
```

Normalize values between 0 and 1

```python
norm = (arr - arr.min()) / (arr.max() - arr.min())
print(np.round(norm, 2))
```

Row and column averages of a 2D array

```python
scores = np.array([[70, 80, 90],
                   [60, 75, 85]])
print(scores.mean(axis=0))   # per column
print(scores.mean(axis=1))   # per row
```

# Pandas

Create a DataFrame and add columns

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Asha", "Ravi", "Meena", "Kiran"],
    "maths": [80, 65, 90, 55],
    "science": [75, 70, 85, 60],
})
df["total"] = df["maths"] + df["science"]
df["average"] = df["total"] / 2
df
```

Filter rows

```python
print(df[df["average"] >= 70])
```

Find the topper

```python
topper = df.loc[df["total"].idxmax(), "name"]
print("Topper:", topper)
```

Pass or Fail using apply

```python
df["result"] = df["average"].apply(lambda x: "Pass" if x >= 60 else "Fail")
print(df[["name", "average", "result"]])
```
