## Introduction to Artificial Intelligence

## What is AI?

Artificial Intelligence (AI) is a technology that enables machines and
computer systems to perform tasks that normally require human intelligence.

AI can perform tasks such as:
- Learning
- Reasoning
- Problem solving
- Decision making
- Natural language understanding
- Image recognition

## Applications of AI

1. Healthcare
2. Banking
3. E-commerce
4. Transportation
5. Education
6. Customer service

## Real-World Examples

- Google Maps route recommendations
- Face recognition in smartphones
- Chatbots
- Fraud detection in banking
- Recommendation systems on shopping platforms

# Introduction to Machine Learning

## What is Machine Learning?

Machine Learning is a subset of Artificial Intelligence that allows
computers to learn patterns from data and make predictions or decisions
without explicitly programming every rule.

## Why is Machine Learning Used?

Machine Learning is used to:

- Make predictions
- Identify patterns
- Automate decisions
- Classify data
- Detect anomalies
- Analyze large amounts of data

## Example

A house price prediction model can learn from historical house data
and predict the price of a new house based on its features.

Types of Machine Learning

You need to learn three main types.

1. Supervised Learning

The training data contains input + known output/label.

Example:

House Size → House Price

The model learns from known examples.

Common algorithms:

Linear Regression
Logistic Regression
Decision Tree
Random Forest
Support Vector Machine

Applications:

House price prediction
Disease classification
Spam detection

2. Unsupervised Learning

The data does not have predefined labels.

The model tries to discover patterns or groups.

Example:

Customer data
     ↓
ML algorithm
     ↓
Customer groups

Common algorithms:

K-Means Clustering
Hierarchical Clustering
PCA

Application:

Grouping customers based on purchasing behavior.

3. Reinforcement Learning

An agent learns by interacting with an environment.

It receives:

Action → Reward/Penalty → Learning

Example:

A robot learns how to navigate a room.

Another famous example is game-playing AI.

Easy way to remember
Supervised      → Learning with labels
Unsupervised    → Finding patterns without labels
Reinforcement   → Learning through rewards

ML Project Lifecycle

This is another important part of your assignment.

The basic lifecycle is:

Data
  ↓
Data Preparation
  ↓
Model Training
  ↓
Evaluation
  ↓
Deployment
  ↓
Monitoring / Improvement
1. Data Collection

Collect data from sources such as:

CSV files
Databases
APIs
Sensors
Websites
2. Data Preparation

Clean and prepare the data.

Examples:

Handle missing values
Remove duplicates
Handle outliers
Encode categorical variables
Scale features
3. Model Training

Give prepared data to an ML algorithm.

Example:

Training Data
     ↓
Random Forest
     ↓
Trained Model
4. Evaluation

Check how well the model performs.

For classification:

Accuracy
Precision
Recall
F1-score

For regression:

MAE
MSE
RMSE
R²
5. Deployment

Make the trained model available for real-world use.

For example:

User
 ↓
Web Application
 ↓
ML Model
 ↓
Prediction
6. Monitoring

After deployment, monitor the model and retrain/update it when necessary.
