## Task 1.20: NumPy practice exercises

```python
# Basic array statistics
import numpy as np
marks = np.array([88, 92, 79, 55, 67, 91, 40])
print("Shape:", marks.shape)
print("Mean :", marks.mean())
print("Max  :", marks.max())
print("Min  :", marks.min())
```

```python
# Filter passing students (marks >= 60)
print(marks[marks >= 60])
```

```python
# Count how many passed and failed
passed = (marks >= 60).sum()
failed = (marks < 60).sum()
print("Passed:", passed, "| Failed:", failed)
```

```python
# Normalize temperatures between 0 and 1
temps = np.array([20, 25, 18, 30, 22])
norm = (temps - temps.min()) / (temps.max() - temps.min())
print(np.round(norm, 2))
```

```python
# Build a scaled identity matrix and get its diagonal
m = np.eye(4) * 5
print(m)
print("Diagonal:", np.diag(m))
```

```python
# Row sums after reshaping
grid = np.arange(1, 21).reshape(4, 5)
print(grid)
print("Row sums:", grid.sum(axis=1))
```

```python
# Replace negative numbers with 0
vals = np.array([5, -3, 8, -1, 0, 12])
vals[vals < 0] = 0
print(vals)
```

```python
# Find the index of the maximum value
sales = np.array([120, 340, 89, 560, 275])
best_day = np.argmax(sales)
print("Best day index:", best_day, "| Value:", sales[best_day])
```

```python
# Sort an array and get top 3 values
prices = np.array([450, 120, 980, 230, 610, 75])
sorted_prices = np.sort(prices)[::-1]
print("Top 3:", sorted_prices[:3])
```

```python
# Element-wise comparison of two arrays
this_week = np.array([200, 150, 300, 180])
last_week = np.array([180, 160, 280, 200])
print("Improved days:", this_week > last_week)
print("Total change:", (this_week - last_week).sum())
```
