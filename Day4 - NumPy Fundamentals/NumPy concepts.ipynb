## Introduction to NumPy

## Why ML uses NumPy
NumPy (Numerical Python) provides the **array**, a grid of same-type values that is far faster
than a Python list for numeric work.

- **Speed:** operations run in optimized C code, not a Python loop.
- **Vectorization:** apply one operation to a whole array at once, no explicit loop needed.
- **Foundation:** Pandas, scikit-learn, TensorFlow and PyTorch are all built on NumPy arrays.
- **Memory efficient:** stores numbers more compactly than a Python list.

```python
import numpy as np
print(np.__version__)
```

```python
# Array Creation (Tasks 1.13, 1.14)
a = np.array([1, 2, 3, 4, 5])          # 1D array
print(a)

b = np.array([[1, 2, 3], [4, 5, 6]])   # 2D array
print(b)
```

```python
zeros = np.zeros((2, 3))
ones = np.ones((3, 3))
identity = np.eye(3)
seq = np.arange(0, 10, 2)         # start, stop, step
lin = np.linspace(0, 1, 5)        # 5 evenly spaced values

print(zeros); print(ones); print(identity); print(seq); print(lin)
```

```python
rng = np.random.default_rng(42)
print(rng.integers(1, 100, 5))
```

```python
# Array properties (1.14)
print("Shape:", b.shape)     # (2, 3)
print("Size :", b.size)      # 6
print("Ndim :", b.ndim)      # 2
print("Dtype:", b.dtype)     # int64
print(np.array([1.5, 2]).dtype)   # float64
```

```python
# Indexing (Task 1.15)
print(a[0])        # first element -> 1
print(a[-1])       # last element  -> 5
print(b[1, 2])      # row 1, col 2 -> 6
print(b[0][1])     # same as b[0, 1] -> 2
```

```python
# Slicing (Task 1.16)
print(a[1:4])    # index 1 to 3 -> [2 3 4]
print(a[:3])     # first 3      -> [1 2 3]
print(a[::2])    # every 2nd    -> [1 3 5]
print(a[::-1])   # reversed     -> [5 4 3 2 1]
```

```python
print(b[:, 1])      # column 1        -> [2 5]
print(b[0, :])      # row 0           -> [1 2 3]
print(b[0:2, 1:3])  # sub-grid        -> [[2 3] [5 6]]
```

```python
print(a[a > 2])    # values greater than 2 -> [3 4 5]
```

```python
# Mathematical Operations (Task 1.17)
x = np.array([10, 20, 30])
y = np.array([1, 2, 3])

print(x + y)     # [11 22 33]
print(x - y)     # [9 18 27]
print(x * y)     # [10 40 90]
print(x / y)     # [10. 10. 10.]
print(x ** 2)    # [100 400 900]
print(x + 5)     # broadcasting: adds 5 to every element -> [15 25 35]
```

```python
# Statistical Operations (Task 1.18)
print("Mean  :", a.mean())     # 3.0
print("Median:", np.median(a)) # 3.0
print("Min   :", a.min())      # 1
print("Max   :", a.max())      # 5
print("Std   :", a.std())      # 1.41
print("Var   :", a.var())      # 2.0
print("Sum   :", a.sum())      # 15
```

```python
# Reshaping Arrays (Task 1.19)
arr = np.arange(1, 13)     # 12 numbers
print(arr.reshape(3, 4))   # 3 rows, 4 columns
print(arr.reshape(4, 3))   # 4 rows, 3 columns
print(arr.reshape(2, 3, 2))  # a 3D array
```
