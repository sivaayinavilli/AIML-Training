## Introduction to Pandas

## What is Pandas?
**Pandas** is a Python library for working with tabular (spreadsheet-like) data.

- **Series:** a single labeled column of data (1D).
- **DataFrame:** a full table of rows and columns (2D), like an Excel sheet.
- Built on top of NumPy, so it's fast, and it adds row/column labels and easy file reading.
- Used for loading, cleaning, filtering, sorting and summarizing data before ML training.

```python
# Series
import pandas as pd

s = pd.Series([10, 20, 30, 40])
print(s)
```

```python
print(s[0])        # 10
print(s[2])        # 30
print(s[1:3])       # index 1 and 2
```

```python
marks = pd.Series([85, 90, 78], index=["Asha", "Ravi", "Meena"])
print(marks)
print(marks["Ravi"])    # 90
```

```python
# DataFrame
data = {
    "name":   ["Asha", "Ravi", "Meena", "Kiran"],
    "age":    [22, 25, 23, 28],
    "dept":   ["IT", "HR", "IT", "Sales"],
    "salary": [30000, 28000, 32000, 40000],
}
df = pd.DataFrame(data)
df
```

```python
df2 = pd.DataFrame([[1, "A"], [2, "B"]], columns=["id", "grade"])
df2
```

```python
print(1+1)
```

```python
df.to_csv("employees.csv", index=False)
df3 = pd.read_csv("employees.csv")
df3
```

```python
print(df3.head(2))      # first 2 rows
print(df3.tail(2))      # last 2 rows
df3.info()               # columns, types, non-null counts
print(df3.describe())   # statistics for numeric columns
```

```python
print("Shape:", df3.shape)
print("Columns:", df3.columns.tolist())
print(df3.dtypes)
```

```python
print(df3["name"])
```

```python
print(df3.loc[1])                       # row with index label 1
print(df3.loc[1, "name"])               # single value -> "Ravi"
print(df3.loc[0:2, ["name", "age"]])    # rows 0-2, chosen columns
```

```python
print(df3.iloc[1])          # 2nd row by position
print(df3.iloc[0:2, 0:2])   # first 2 rows, first 2 columns
print(df3.iloc[-1])         # last row
```

```python
print(df3[df3["salary"] > 28000])
```

```python
print(df3[(df3["dept"] == "IT") & (df3["age"] > 22)])
```

```python
print(df3[df3["dept"].isin(["IT", "HR"])])
```

```python
# Sorting
print(df3.sort_values("salary"))                       # ascending
print(df3.sort_values("salary", ascending=False))      # descending
print(df3.sort_values(["dept", "salary"]))              # sort by two columns
```
