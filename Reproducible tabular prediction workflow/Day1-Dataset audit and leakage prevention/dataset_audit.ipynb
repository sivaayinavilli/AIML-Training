```python
import pandas as pd

df = pd.read_csv("breast_cancer.csv")

print("Dataset shape:", df.shape)
print("\nMissing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("\nTarget distribution:")
print(df["target"].value_counts().sort_index())
```

```python
# Separate features and target
X = df.drop(columns=["target"])
y = df["target"]

print("Number of records:", len(df))
print("Number of features:", X.shape[1])
print("Target column:", y.name)

print("\nTarget distribution:")
print(y.value_counts().sort_index())

print("\nTarget percentages:")
print((y.value_counts(normalize=True).sort_index() * 100).round(2))

print("\nFeature column names:")
print(X.columns.tolist())
```

```python
# Check whether any feature is an exact copy of the target
matching_features = []

for column in X.columns:
    if X[column].equals(y):
        matching_features.append(column)

print("Features identical to target:", matching_features)

if not matching_features:
    print("No feature is an exact copy of the target.")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target counts:")
print(y_train.value_counts().sort_index())

print("\nTesting target counts:")
print(y_test.value_counts().sort_index())
```

```python
X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training split reproducible:",
      X_train.equals(X_train_2))

print("Testing split reproducible:",
      X_test.equals(X_test_2))

print("Training labels reproducible:",
      y_train.equals(y_train_2))

print("Testing labels reproducible:",
      y_test.equals(y_test_2))
```

```python
from pathlib import Path
import pandas as pd

reports_dir = Path("../reports")
reports_dir.mkdir(parents=True, exist_ok=True)

split_summary = pd.DataFrame({
    "split": ["train", "test"],
    "records": [len(X_train), len(X_test)],
    "malignant": [
        int((y_train == 0).sum()),
        int((y_test == 0).sum())
    ],
    "benign": [
        int((y_train == 1).sum()),
        int((y_test == 1).sum())
    ]
})

split_summary.to_csv(
    reports_dir / "baseline_split_summary.csv",
    index=False
)

display(split_summary)
```
